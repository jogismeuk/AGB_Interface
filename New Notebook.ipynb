```python
import arcpy

fc = r"C:\ArcProj\AboveGroundBiomass\AboveGroundBiomass.gdb\WoodyBiomassPolygonsCopy_ExportFeatures"
agb_field = "AGB"        # your biomass in kg
out_field = "AGB_T_HA"      # output t/ha

# Add output field if missing
fields = [f.name for f in arcpy.ListFields(fc)]
if out_field not in fields:
    arcpy.management.AddField(fc, out_field, "DOUBLE")

# t/ha = (AGB_kg * 10) / area_m2
expr = f"(!{agb_field}! * 10.0) / !shape.area@SQUAREMETERS!"
arcpy.management.CalculateField(fc, out_field, expr, "PYTHON3")
```
